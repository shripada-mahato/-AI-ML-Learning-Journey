```python
def power(number, exponent=2):
    return number ** exponent
print(power(2,4))
```

```python
nums = [2, 7, 11, 15]
target = 9
for i in range(len(nums)-1):
    if nums[i] + nums[i+1] == target:
        print(i,i+1)
```

```python
def palimdrome(name):
    if name == name[::-1]:
        return True
    else:
        return False
print(palimdrome("mam"))
```
