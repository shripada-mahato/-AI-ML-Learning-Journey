```python
def calculate(**kwargs):
  sum = 0
  for i in kwargs.values():
    sum += i
  return sum
print(calculate(a=10,b=20,c=30))
```

```python
def calculate_average(**kwargs):
  total = 0
  count = 0
  for i in kwargs.values():
    total += i
    count += 1
  return total / count
print(calculate_average(
    python=80,
    java=70,
    sql=90,
    c=100
))

```

```python
def count_positive_negative(arr):
  positive = 0
  negative = 0
  zero = 0
  for i in arr:
    if i > 0:
      positive += 1
    elif i < 0:
      negative += 1
    else:
      zero += 1
  return positive, negative, zero
print(count_positive_negative([10,-5,0,8,-2,7]))
```

```python
def find_first_negative(arr):
  negative = 0
  for i in arr:
    if i < 0:
      negative = arr[i]
      break
  return negative
find_first_negative([4,7,-3,-8,2])
```
