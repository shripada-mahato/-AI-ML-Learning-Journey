```python
def power(number, exponent=2):
    return number ** exponent
print(power(2,4))
```

```python
nums = [2, 7, 11, 15]
target = 9
for i in range(len(nums)-1):
    if nums[i] + nums[i+1] == target:
        print(i,i+1)
```

```python
def palimdrome(name):
    if name == name[::-1]:
        return True
    else:
        return False
print(palimdrome("mam"))
```

```python
def calculate_price(price,discount=10):
  offer = price * discount / 100
  return price - offer
print(calculate_price(1000,20))
```

```python
def multiply_number(*args):
  total = 1
  for i in args:
    total *= i
  return total
print(multiply_number(2,3,4,5))
```

```python
def find_largest(*args):
  largest = args[0]
  for i in args:
    if i > largest:
      largest = i
  return largest
print(find_largest(-5,-8,-2,9,7))
```

```python
def count_even(*args):
  count = 0
  for i in args:
    if i % 2 == 0:
      count += 1
  return count
print(count_even(5,8,7,6,9,1,2))
count_even(-4,-3,0,2,7,10)
```

```python
def student_details(**kwargs):
  for key, value in kwargs.items():
    print(key,":",value)
student_details(name='shri',
                course='bca',
                college='kolkata',
                year=3)
```

```python
def calculate_marks(**kwargs):
  total_marks = 0
  for key, value in kwargs.items():
    total_marks += value
  return total_marks
calculate_marks(python=50,
                java=40,
                cpp=10)
```

```python
def find_highest_marks(**kwargs):
  highest_marks = None
  highest_subject = None

  for key, value in kwargs.items():
    if highest_marks is None or value > highest_marks:

      highest_marks = value
      highest_subject = key

  return highest_subject, highest_marks

find_highest_marks(python=50,
                   java=92,
                   c=98)
```

```python
def find_lowest_marks(**kwargs):
  lowest_subject = None
  lowest_marks = None

  for key, value in kwargs.items():
    if lowest_marks == None or value < lowest_marks:
      lowest_subject = key
      lowest_marks = value

    return lowest_subject, lowest_marks
find_lowest_marks(python=85,
                  java=78,
                  sql=21,
                  cpp=12)
```
