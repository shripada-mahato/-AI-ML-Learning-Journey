```
import pandas as pd

df = pd.DataFrame({
    "Product_ID": [101, 102, 103, 104, 105, 106],
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer", "Webcam"],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics", "Electronics", "Accessories"],
    "Price": [55000, 800, 1500, 12000, 9000, 2500],
    "Stock": [10, 50, 30, 15, 8, 25],
    "Rating": [4.5, 4.2, 4.6, 4.1, 3.9, 4.4]
})
print("Original Dataset:- \n",df)
rmv_rting_clm = df.drop('Rating',axis=1)
print("Remove rating column from the dataset:- \n",rmv_rting_clm)
rmv_stck_rting_clm = df.drop(['Stock','Rating'],axis=1)
print("Remove stock and rating column:- \n",rmv_stck_rting_clm)
rmv_indx_3 = df.drop(3,axis=0)
print("Remove the row with index 3:- \n",rmv_indx_3)
rmv_indx_1_4 = df.drop([1,4],axis=0)
print("Remove rows with indexes 1 and 4:- \n",rmv_indx_1_4)
clean_df = df.drop(['Rating','Stock'],axis=1).drop([2,5],axis=0)
print(clean_df)
```
