```python
from matplotlib.lines import lineStyles
import matplotlib.pyplot as plt
month = ['jan','feb','mar','apr','may','jun']
traffic = [10,15,20,20,60,30]
plt.figure(figsize=(8,3))
plt.plot(
    month,traffic,
    label='month',
    color='green',
    linestyle='--',
    marker='o',
    markeredgecolor='yellow'

)
plt.grid()
plt.legend()
plt.title('Month Wise Traffice Analysis')
plt.xlabel('month')
plt.ylabel('traffic')
plt.show()
```
