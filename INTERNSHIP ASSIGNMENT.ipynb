TASK-1 NUMPY OPERATION

```python
import pandas as pd
import numpy as np
df = pd.read_csv("/content/students_dataset.csv")
print(df)
mean_each_sub = df.groupby('Student')[['Math','Science','English']].agg(['mean','min','max'])
print(mean_each_sub)
df['Total_Marks']=df['Math']+df['Science']+df['English']
df['Percentage % ']=df['Total_Marks']*100/300
df['Status']= np.where(df['Attendance']>80,'Eligilbe','Not Eligible')
print(df.head().to_string())
```

PANDAS TASK-2

```python
print(df.head(10).to_string())
print(df.tail(5).to_string())
print("Shape of the Dataset:-",df.shape)
print("Duplicate Value:- \n",df.duplicated())
df['Result_Status']=np.where((df['Total_Marks']/3)>70,'pass','fail')
print("Result Status Pass Fail top 10 row :- \n",df.head(10).to_string())
print("Result Status Pass Fail last 10 row :- \n",df.tail(10).to_string())
```
