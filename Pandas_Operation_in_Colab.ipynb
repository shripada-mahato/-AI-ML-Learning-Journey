```
import pandas as pd

df = pd.DataFrame({
    "Product_ID": [101, 102, 103, 104, 105, 106],
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer", "Webcam"],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics", "Electronics", "Accessories"],
    "Price": [55000, 800, 1500, 12000, 9000, 2500],
    "Stock": [10, 50, 30, 15, 8, 25],
    "Rating": [4.5, 4.2, 4.6, 4.1, 3.9, 4.4]
})
print("Original Dataset:- \n",df)
rmv_rting_clm = df.drop('Rating',axis=1)
print("Remove rating column from the dataset:- \n",rmv_rting_clm)
rmv_stck_rting_clm = df.drop(['Stock','Rating'],axis=1)
print("Remove stock and rating column:- \n",rmv_stck_rting_clm)
rmv_indx_3 = df.drop(3,axis=0)
print("Remove the row with index 3:- \n",rmv_indx_3)
rmv_indx_1_4 = df.drop([1,4],axis=0)
print("Remove rows with indexes 1 and 4:- \n",rmv_indx_1_4)
clean_df = df.drop(['Rating','Stock'],axis=1).drop([2,5],axis=0)
print(clean_df)
```

```
import pandas as pd

df = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Rohit", "Anjali", "Karan", "Neha"],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Marketing", "Finance", "IT"],
    "City": ["Delhi", "Mumbai", "Delhi", "Kolkata", "Mumbai", "Delhi", "Pune", "Chennai"],
    "Salary": [35000, 50000, 42000, 55000, 38000, 60000, 45000, 52000],
    "Experience": [2, 5, 3, 6, 2, 7, 4, 5]
})
print(df)
print(df[['Name','Salary']])
print(df[df['Salary']>45000])
print(df[(df['Department']=='IT') & (df['Salary']>50000)])
df['Salary'].accending=True
salary_sort=df.sort_index(ascending=False)
print(salary_sort)
print(salary_sort.head(3))
df['bonus']=df['Salary']/10
print(df)
df['Total_Salary']=df['Salary']+df['bonus']
print(df.to_string())
ave_salary = df.groupby('Department')['Salary'].mean()
print(ave_salary)
max_exp = df.groupby('Department')['Experience'].max()
print(max_exp)
agg_in = df.groupby('Department')['Salary'].agg(['mean','max','min','sum'])
print(agg_in)
print(df['City'].unique())
print(df['City'].count())
print(df['City'].value_counts())
clean_df = df.drop(['Experience'],axis=1).drop([1,5],axis=0)
print(clean_df)
df.rename({'Emp_ID':'Employee_ID'})
df.rename({'Salary':'Annual_Salary'})
df.rename({'Experience':'Exp_Years'})
print(df)
```
